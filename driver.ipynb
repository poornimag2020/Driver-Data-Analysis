```python
data[
    (data["mean_dist_day"] > 100) &
    (data["mean_over_speed_perc"] > 50)
]
```

```python
data["speed_category"] = pd.cut(
    data["mean_over_speed_perc"],
    bins=[-1, 5, 10, 20, 50, 100],
    labels=["0-5", "6-10", "11-20", "21-50", "51-100"]
)
```

```python
data["speed_category"].value_counts()
```

```python
data.groupby(
    "speed_category",
    observed=True
)["mean_dist_day"].mean()
```

```python
data.groupby(
    "speed_category",
    observed=True
)["mean_dist_day"].max()
```

```python
data.groupby(
    "speed_category",
    observed=True
).agg({
    "mean_dist_day": ["mean", "max", "min"],
    "mean_over_speed_perc": ["mean", "max", "min"]
})
```

```python
data[[
    "mean_dist_day",
    "mean_over_speed_perc"
]].corr()
```

```python
data["Distance_Category"] = np.where(
    data["mean_dist_day"] > 100,
    "High Distance",
    "Low Distance"
)
```
